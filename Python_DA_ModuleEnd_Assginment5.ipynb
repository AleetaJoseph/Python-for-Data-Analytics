❖	Task 2 — Data Cleaning

●	 Cleaning Missing Data

■	Detect missing values (isnull(), isna())


```python
df.isnull().sum()
```

```python
df.isna().sum()
```

```python
df.describe()
```

```python
#■	Handle using: dropna(), fillna(), median/mode, forward/backward-fill
df["age"]= df["age"].fillna(38.000000)
df["likes"]= df["likes"].fillna(10105.500000)
df["comments"]= df["comments"].fillna(1497.000000)
df["shares"]= df["shares"].fillna(1012.000000)
```

```python
df["gender"].value_counts()
```

```python
df["sentiment"].value_counts()
```

```python
df["gender"]=df["gender"].fillna("Male")
df["sentiment"]=df["sentiment"].fillna("Positive")
```

```python
#Identify & remove duplicates
df.duplicated().sum()

#No duplicates found, so no need to remove any duplicates
```

```python
#Data Formatting
#	Fix incorrect data types

df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["comments"] = pd.to_numeric(df["comments"], errors="coerce")
df["shares"] = pd.to_numeric(df["shares"], errors="coerce")
```

```python
df.dtypes
```

```python
df.head(3)
```

```python
#	Standardize categories (e.g., gender labels)
print(df["gender"].unique())
print(df["country"].unique())
print(df["post_type"].unique())
print(df["post_category"].unique())
print(df["device_type"].unique())
print(df["sentiment"].unique())
```

```python
##The sentiment column contained both positive and Positive, so the values were standardized to lowercase. The remaining categorical columns 
# were already consistently formatted.
df["sentiment"] = df["sentiment"].str.lower()
```

```python
df["sentiment"].unique()
```

```python
#Correct unrealistic values in likes, comments, shares
df[["likes", "comments", "shares"]].min()

#no unrealistic values found in likes, comments, shares columns. All values are greater than 0.
```
